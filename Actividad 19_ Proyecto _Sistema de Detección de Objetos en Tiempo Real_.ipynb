```python
import cv2
from ultralytics import YOLO

# 1. Cargar el modelo YOLO optimizado y ligero
model = YOLO("yolov8n.pt") 

# 2. Abrir la cámara web local (índice 0)
cap = cv2.VideoCapture(0)

print("Iniciando transmisión en tiempo real. Presiona la letra 'q' en tu teclado para salir.")

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        print("No se pudo acceder a la cámara.")
        break

    # Realizar la detección de objetos en el fotograma actual
    results = model(frame, stream=True)

    # Dibujar las cajas y etiquetas de múltiples objetos detectados sobre el video
    for r in results:
        annotated_frame = r.plot()

    # Mostrar el video resultante en una ventana
    cv2.imshow("Deteccion de Objetos - YOLO", annotated_frame)

    # Romper el bucle si se presiona la tecla 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

# Liberar la cámara y cerrar ventanas
cap.release()
cv2.destroyAllWindows()
```
