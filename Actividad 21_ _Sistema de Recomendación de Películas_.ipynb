```python
import pandas as pd
import numpy as np
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 1. Agrupar usuarios y Simular datos
def agrupar_usuarios_streaming():
    """Simula perfiles de usuarios y aplica clustering por gustos cinematográficos"""
    print("=== 1. Agrupación de Usuarios (Clustering) ===")
    
    # datos simulados: 300 usuarios con calificaciones en 5 géneros de películas
    # Géneros: [Acción, Comedia, Drama, Ciencia Ficción, Terror]
    np.random.seed(42)
    n_usuarios = 300
    
    # Creamos 3 perfiles principales de usuarios para simular grupos naturales
    grupo_1 = np.random.normal(loc=[8, 2, 2, 7, 3], scale=1.5, size=(100, 5)) # Fans de Acción y Sci-Fi
    grupo_2 = np.random.normal(loc=[2, 8, 7, 2, 2], scale=1.5, size=(100, 5)) # Fans de Comedia y Drama
    grupo_3 = np.random.normal(loc=[3, 3, 2, 2, 8], scale=1.5, size=(100, 5)) # Fans de Terror
    
    X = np.vstack([grupo_1, grupo_2, grupo_3])
    # Aseguramos que las calificaciones estén entre 0 y 10
    X = np.clip(X, 0, 10)
    
    generos = ['Acción', 'Comedia', 'Drama', 'Ciencia Ficción', 'Terror']
    
    # Aplicar K-Means para encontrar 3 clústeres de usuarios
    kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
    clusters = kmeans.fit_predict(X)
    
    print(f"Total de usuarios analizados: {n_usuarios}")
    print(f"Grupos (Clústeres) identificados: 3")
    print(f"Distribución de usuarios por grupo: {np.bincount(clusters)}\n")
    
    return X, clusters, kmeans, generos

# 2. Sistema de recomendación
def recomendar_peliculas(usuario_id, X, clusters, generos):
    """Genera recomendaciones basadas en el grupo al que pertenece el usuario"""
    print("=== 2. Sistema de Recomendación ===")
    
    # Catálogo simulado de películas por género favorito
    catalogo = {
        'Acción': ['Matrix', 'Gladiador', 'John Wick'],
        'Comedia': ['Superbad', 'The Mask', 'Grown Ups'],
        'Drama': ['The Godfather', 'Forrest Gump', 'Titanic'],
        'Ciencia Ficción': ['Interstellar', 'Blade Runner', 'Inception'],
        'Terror': ['El Conjuro', 'It', 'Hereditary']
    }
    
    # Identificar el grupo del usuario objetivo
    grupo_usuario = clusters[usuario_id]
    preferencias_usuario = X[usuario_id]
    
    print(f"Perfil del usuario #{usuario_id} (Perteneciente al Clúster {grupo_usuario}):")
    for g, pref in zip(generos, preferencias_usuario):
        print(f" - {g}: {pref:.1f}/10")
        
    # Encontrar el género con mayor puntaje para este usuario
    genero_favorito = generos[np.argmax(preferencias_usuario)]
    recomendaciones = catalogo[genero_favorito]
    
    print(f"\nPelículas recomendadas para el usuario #{usuario_id} (basado en su gusto por '{genero_favorito}'):")
    for pelicula in recomendaciones:
        print(f"   -> {pelicula}")
    print("-" * 40)

# 3. Evaluar efectividad
def evaluar_efectividad(X, clusters):
    """Mide la calidad de la agrupación usando el coeficiente de Silueta"""
    print("=== 3. Evaluación de Efectividad ===")
    
    # El Silhouette Score mide qué tan bien agrupados están los usuarios (-1 a 1)
    # Un valor cercano a 1 indica que los grupos están bien separados y definidos.
    score = silhouette_score(X, clusters)
    
    print(f"Puntaje de Silueta del Clustering: {score:.4f}")
    if score > 0.5:
        print("Evaluación: Excelente separación de gustos. Los grupos de usuarios están bien diferenciados.")
    else:
        print("Evaluación: Los grupos se solapan un poco, se recomienda ajustar las características.")
    print("-" * 40)

# --- Ejecución principal del script ---
if __name__ == "__main__":
    X, clusters, kmeans, generos = agrupar_usuarios_streaming()
    recomendar_peliculas(usuario_id=15, X=X, clusters=clusters, generos=generos)
    evaluar_efectividad(X, clusters)
```
