```python
import random

class EnjambreRescate:
    """Clase central que coordina la red de comunicación y el mapa compartido del enjambre"""
    def __init__(self, ancho, alto):
        self.ancho = ancho
        self.alto = alto
        self.mapa_explorado = set()
        self.victimas_encontradas = []
        self.buzon_mensajes = []

    def enviar_mensaje(self, emisor, tipo, datos):
        """Sistema de comunicación: permite a un robot transmitir alertas al enjambre"""
        mensaje = {"emisor": emisor, "tipo": tipo, "datos": datos}
        self.buzon_mensajes.append(mensaje)
        print(f"📡 [RED DE COMUNICACIÓN] {emisor} transmitió alerta -> [{tipo}] en coordenadas {datos}")

    def mostrar_historial_mensajes(self):
        print(f"\n📬 Total de transmisiones en la red: {len(self.buzon_mensajes)}")
        for msg in self.buzon_mensajes:
            print(f"   • De {msg['emisor']}: {msg['tipo']} -> {msg['datos']}")


class RobotRescate:
    def __init__(self, id_robot, enjambre):
        self.id = f"Robot-{id_robot}"
        self.enjambre = enjambre
        self.posicion_actual = (0, 0)
        self.victimas_rescatadas = 0

    def navegar_a(self, x, y):
        """Implementa la navegación hacia un punto específico de su sector"""
        self.posicion_actual = (x, y)
        print(f"🤖 [{self.id}] Se desplaza a la posición ({x}, {y})")

    def ejecutar_busqueda_en_zona(self, zona_asignada):
        """Estrategia colectiva: explora de forma eficiente la zona asignada"""
        print(f"\n🚀 [{self.id}] Iniciando patrullaje en su sector asignado...")
        
        for x, y in zona_asignada:
            # Evita re-explorar áreas ya cubiertas por otros compañeros del enjambre
            if (x, y) in self.enjambre.mapa_explorado:
                continue

            self.navegar_a(x, y)
            self.enjambre.mapa_explorado.add((x, y))

            # Simular detección aleatoria de una víctima (15% de probabilidad)
            if random.random() < 0.15:
                print(f"🚨 ¡ALERTA! [{self.id}] ha localizado una VÍCTIMA en ({x}, {y})!")
                self.victimas_rescatadas += 1
                self.enjambre.victimas_encontradas.append((x, y))
                
                # Comunicación: Notificar a todo el enjambre
                self.enjambre.enviar_mensaje(self.id, "VICTIMA_LOCALIZADA", (x, y))

            # Simular detección de un obstáculo peligroso (10% de probabilidad)
            elif random.random() < 0.10:
                print(f"⚠️ [{self.id}] Obstáculo infranqueable detectado en ({x}, {y}).")
                self.enjambre.enviar_mensaje(self.id, "OBSTACULO_DETECTADO", (x, y))


def simular_enjambre_rescate():
    print("🚁 --- INICIANDO SIMULADOR DE ENJAMBRE DE BÚSQUEDA Y RESCATE ---")
    
    # Definir el tamaño del área de desastre (cuadrícula de 6x6)
    ancho, alto = 6, 6
    sistema_enjambre = EnjambreRescate(ancho, alto)

    # 1. COORDINACIÓN Y ESTRATEGIA COLECTIVA: Dividir el mapa en sectores (2 robots)
    # Robot 1 toma la mitad izquierda, Robot 2 toma la mitad derecha
    mitad_ancho = ancho // 2
    zona_robot_1 = [(x, y) for y in range(alto) for x in range(mitad_ancho)]
    zona_robot_2 = [(x, y) for y in range(alto) for x in range(mitad_ancho, ancho)]

    robot_1 = RobotRescate(1, sistema_enjambre)
    robot_2 = RobotRescate(2, sistema_enjambre)

    # Simular la ejecución de búsqueda simultánea o por turnos coordinados
    max_pasos = max(len(zona_robot_1), len(zona_robot_2))
    for paso in range(max_pasos):
        if paso < len(zona_robot_1):
            robot_1.ejecutar_busqueda_en_zona([zona_robot_1[paso]])
        if paso < len(zona_robot_2):
            robot_2.ejecutar_busqueda_en_zona([zona_robot_2[paso]])

    # Resultados y Resumen del Enjambre
    print("\n" + "="*50)
    print("📊 REPORTE FINAL DE LA MISIÓN DEL ENJAMBRE")
    print("="*50)
    print(f"🗺️ Área total cubierta: {len(sistema_enjambre.mapa_explorado)} de {ancho*alto} celdas.")
    print(f"👥 Total de víctimas rescatadas por el equipo: {len(sistema_enjambre.victimas_encontradas)}")
    print(f"🔴 Víctimas ubicadas en coordenadas: {sistema_enjambre.victimas_encontradas}")
    
    sistema_enjambre.mostrar_historial_mensajes()

if __name__ == "__main__":
    simular_enjambre_rescate()
```
